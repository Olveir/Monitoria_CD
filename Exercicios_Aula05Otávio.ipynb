# Gráficos

> A partir do que foi visto em sala, faça os seguintes gráficos:

- Distribuição da quantidade de sobreviventes.
- Sexo em relação a quantidade de sobreviventes.
- Boxplot para destribuição de sobreviventes por idade.
- Histograma de idade.

> Faça uma pequena análise para cada gráfico com as informações relevantes obtidas através das visualizações. E caso encontre, aponte possíveis outliers/anomalias.

```python
plt.figure(figsize=(8, 6))
sns.countplot(x='survived', data=df, hue='survived', order=df['survived'].value_counts().index)
plt.xlabel('Passageiros')
plt.ylabel('Quantidade de Sobreviventes')
plt.title('Distribuição da quantidade de sobreviventes.')
plt.legend(title='Não Sobreviveu: 0\nSobreviveu: 1')
plt.show()
```

```python
sobreviventesSex = df[['survived','sex']].groupby('sex').sum()
sobreviventesSex

sns.barplot(x='sex', y = 'survived', hue='survived', data=sobreviventesSex)
plt.title('Distribuição de sobreviventes por sexo')
plt.xlabel('Sexo')
plt.ylabel('Quantidade de sobreviventes')
plt.show()
```

```python
#Boxplot para destribuição de sobreviventes por idade.

sns.boxplot(y='age',x ='survived', data=df, hue='survived')
plt.title('Relação de sobreviventes por idade.')
plt.xlabel('Sobreviventes')
plt.ylabel('Idade')
plt.legend(title='Não Sobreviveu: 0\nSobreviveu: 1',loc='upper right')
plt.show()
```

```python
#Histograma de idade.

plt.figure(figsize=(8, 6))
sns.histplot(df['age'], kde=True)
plt.xlabel('Idade')
plt.ylabel('Frequência')
plt.title('Distribuição da idade dos passageiros')
plt.show()
```
